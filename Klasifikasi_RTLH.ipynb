```python
# Import library
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

from google.colab import files

print("Library berhasil diimport")
```

```python
uploaded = files.upload()

nama_file = list(uploaded.keys())[0]

df = pd.read_csv(nama_file, sep=';')

df.head()
```

```python
print("ukuran Dataset:", df.shape)

print("\nNama Kolom:")
print(df.columns.tolist())

print("\nInformasi Dataset:")
df.info()

display(df.head())
```

```python
display(df.describe(include='all'))
```

```python
df.columns = (
    df.columns
    .str.strip()
    .str.lower()
    .str.replace(' ','_')
)
print(df.columns.tolist())
```

```python
print("Jumlah Missing Value:")
print(df.isnull().sum())
```

```python
print("Data duplikat:", df.duplicated().sum())
df = df.drop_duplicates()

print("jumlah data setelah menghapus duplikat:", df.shape)
```

```python
df = df.dropna(axis=1, how='all')

df = df.dropna(axis=0, how='all')

print("ukuran data setelah pembersihan:",
      df.shape)
```

```python
for kolom in df.select_dtypes(include='object').columns:
    df[kolom] = df[kolom].str.strip()

print("Pembersihan teks selesai")
display(df.head())
```

```python
for kolom in df.columns:
    if df[kolom].dtype == 'object':

        modus = df[kolom].mode()
        if not modus.empty:
            df[kolom] = df[kolom].fillna(modus[0])
    else:

        df[kolom] = df[kolom].fillna(df[kolom].median())

print("Missing value setelah preprocessing:")
print(df.isnull().sum())
```

```python
display(df.head())

df.info()

print("Total missing value:",
      df.isnull().sum().sum())

print("Total data duplikat:",
      df.duplicated().sum())

print("Ukuran akhir dataset:", df.shape)
```

```python
df.to_csv('dataset_RTLH_preprocessing.csv',
          index=False
)

files.download('dataset_RTLH_preprocessing.csv')
```
