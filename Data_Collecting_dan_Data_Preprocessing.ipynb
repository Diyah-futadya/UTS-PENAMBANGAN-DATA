```python
import pandas as pd
import numpy as np

from google.colab import files
```

```python
uploaded = files.upload()

nama_file = list(uploaded.keys())[0]

df = pd.read_csv(nama_file)

display(df.head())
```

```python

df = df.iloc[:, :9]


df.columns = [
    "Provinsi",
    "Kabupaten",
    "Tahun",
    "Kode_Referensi",
    "Kecamatan",
    "Jumlah_Rumah",
    "Jumlah_RLH",
    "Jumlah_RTLH",
    "Satuan"
]

display(df.head())
```

```python
df = df.dropna(how="all")


df = df[
    df["Kecamatan"].notna()
]

df = df[
    df["Kecamatan"].astype(str).str.lower() != "jumlah"
]

df = df.reset_index(drop=True)

print("Jumlah data:", len(df))

display(df)
```

```python
kolom_numerik = [
    "Tahun",
    "Jumlah_Rumah",
    "Jumlah_RLH",
    "Jumlah_RTLH"
]

for kolom in kolom_numerik:
    df[kolom] = pd.to_numeric(
        df[kolom], errors="coerce"
    )

print(df.dtypes)
```

```python
print("Data kosong:")
print(df.isnull().sum())

print("\nData duplikat:")
print(df.duplicated().sum())
```

```python
df["Persentase_RTLH"] = (
    df["Jumlah_RTLH"] /
    df["Jumlah_Rumah"]
) * 100

display(
    df[[
        "Kecamatan",
        "Jumlah_Rumah",
        "Jumlah_RTLH",
        "Persentase_RTLH"
    ]]
)
```

```python
def kategori_rtlh(nilai):
    if nilai < 1:
        return "Rendah"
    elif nilai < 3:
        return "Sedang"
    else:
        return "Tinggi"

df["Kategori_RTLH"] = (
    df["Persentase_RTLH"].apply(kategori_rtlh)
)

display(
    df[[
        "Kecamatan",
        "Persentase_RTLH",
        "Kategori_RTLH"
    ]]
)
```

```python
print("Distribusi Kategori RTLH:")

print(
    df["Kategori_RTLH"].value_counts()
)
```

```python
df.to_csv(
    "dataset_rtlh_nganjuk_preprocessing.csv",
    index=False
)

files.download(
    "dataset_rtlh_nganjuk_preprocessing.csv"
)
```
