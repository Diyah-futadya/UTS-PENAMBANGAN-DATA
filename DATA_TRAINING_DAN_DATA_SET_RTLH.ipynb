```python
import pandas as pd
import numpy as np

from sklearn.model_selection import train_test_split

print("Library berhasil diimport")
```

```python
df = pd.read_csv('dataset_RTLH_preprocessing.csv')

display(df.head())

print("Jumlah baris dan kolom:",
      df.shape)
```

```python
target = 'status'

if target not in df.columns:
    raise ValueError(
        "Kolom status belum tersedia. "
        "Tambahkan label hasil verifikasi RTLH terlebih dahulu"
    )
x = df.drop(columns=[target])
y = df[target]

print("Atribut prediktor:")
print(x.columns.tolist())

print("\nDistribusi kelas:")
print(y.value_counts())
```

```python
x_train, x_test, y_train, y_test = train_test_split(
    x, y, test_size=0.2, random_state=42, stratify=y
)
print("Jumlah data training:", len(x_train))
print("Jumlah data testing:", len(x_test))

print("\nPersentase data training:", len(x_train)/len(x)*100, "%")
print("Persentase data testing:", len(x_test)/len(x)*100, "%")
```
